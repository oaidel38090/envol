# Introduction à Docker

# Docker compose

`docker compose` est un outil fourni par Docker qui simplifie le processus de gestion de plusieurs conteneurs. 
Plutôt que de gérer individuellement chaque conteneur, `docker compose` permet de gérer plusieurs conteneurs simultanément à travers un fichier unique appelé docker-compose.yml. 
Ce fichier définit toutes les configurations stockage, réseau et applicatifs nécessaires au démarrage d'une application que vous souhaitez exécuter.
On peut même exprimer des dépendances entre les conteneurs. 

Vous pouvez donc déployer un ensemble de services connectés et interdépendants, comme une application web et sa base de données, avec des commandes simples (docker-compose up ...). 
Dans ce module, un fichier `docker-compose.yml` est disponible dans `$monrepertoire/docker` et il est personnalisé pour le besoin du module.

# Tester votre environnement

## Tester neo4j

1 - A partir de votre terminal, positionner vous dans le répertoire `$monrepertoire/docker` précédemment créé :

2 - Executer la commande

```
docker-compose up -d neo4j
```

---

**NOTE**

La commande `docker-compose up ... ` est l'équivalent d'un docker run, elle permet d'instancier un environnement pour l'exécution de neo4j. On utilisera la commande `docker stop neo4j` ou `docker start neo4j` par la suite.

---


3 - Vérifiez que l'application neo4j est bien démarré

```[bigdata@oaldum02:~]$ docker ps
CONTAINER ID   IMAGE     COMMAND                  CREATED         STATUS          PORTS                                                                                            NAMES
bca590b93fca   neo4j     "tini -g -- /startup…"   5 minutes ago   Up 12 seconds   0.0.0.0:7474->7474/tcp, :::7474->7474/tcp, 7473/tcp, 0.0.0.0:7687->7687/tcp, :::7687->7687/tcp   neo4j
```

4 - Connectez-vous à neo4j :

Pour les TP neo4j, nous utiliserons l'interface web fournit avec neo4j pour interagir avec celui-ci. Une fois la container neo4j démarré accéder à l'url suivante :

[http://127.0.0.1:7474/browser/](http://127.0.0.1:7474/browser/)

Les identifiants de connexion à neo4j sont les suivants : login : neo4j pwd : neo4j

Lors de votre première connexion, vous devrez changer le mot de passe par "nosql2024" .

Votre environnement neo4j est à présent opérationnel.

5 - Arrêter le container neo4j

```
docker stop neo4j
```

Votre environnement neo4j fonctionne correctement.

## Tester mongoDB

1 - A partir de votre terminal, positionner vous dans le répertoire `$monrepertoire/docker` précédemment créé :

2 - Executer la commande

```
docker-compose up -d mongosingle
```



3 - Vérifier que le container mongo est bien démarré

```
[bigdata@oaldum02 ~]$ docker ps
CONTAINER ID   IMAGE                      COMMAND                  CREATED             STATUS                       PORTS                      NAMES
adda865812bf   mongo:4.4.6                "docker-entrypoint.s…"   7 seconds ago       Up 6 seconds                 0.0.0.0:27017->27017/tcp   mongosingle
```

4 - Connectez-vous au container `mongosingle` :

```
docker exec -it mongosingle bash 
```

5 - Utiliser le client (mongo) embarqué dans le container pour vous connecter au service mongo :

```
root@9c93312e4658:/# mongo
>
> show dbs
admin   0.000GB
config  0.000GB
local   0.000GB
>exit
bye
root@9c93312e4658:/# exit
exit
```

6 - Arrêter le container mongo

```
docker stop mongosingle
```

Vous êtes à présent pour la formation.
